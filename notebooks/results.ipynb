````python
%matplotlib inline
import sys
from pathlib import Path

# Make the project importable whether Jupyter was started here or in the project folder.
root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(root))

import pandas as pd
from IPython.display import display

from remittrend import artifacts, config, report

pd.set_option('display.max_rows', 120)
pd.set_option('display.float_format', '{:,.3f}'.format)

data = report.load()
tables = report.tables(data)
figures = report.figures(data)
monthly = data['monthly']
print(f"Study period: {monthly.index[0]} to {monthly.index[-1]}, {len(monthly)} months")
````
